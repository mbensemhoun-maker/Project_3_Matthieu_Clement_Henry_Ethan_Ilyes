## CSV Creation


```python
from pathlib import Path
import pandas as pd

# Find the workspace root from the notebook's current working directory.
workspace = next(
    (parent for parent in [Path.cwd(), *Path.cwd().parents]
     if (parent / "cv_scoring_ai_V1" / "datas_extract").is_dir()),
    None,
)
if workspace is None:
    raise FileNotFoundError("Could not find cv_scoring_ai_V1/datas_extract from the current directory")

source_dir = workspace / "cv_scoring_ai_V1" / "datas_extract"
output_csv = workspace / "cv_scoring_ai_V2" / "data" / "datas_extract.csv"
rows = [
    {"Name": path.name, "Content": path.read_text(encoding="utf-8")}
    for path in sorted(source_dir.glob("*.txt"))
]
df_txt = pd.DataFrame(rows, columns=["Name", "Content"])
df_txt.to_csv(output_csv, index=False, encoding="utf-8-sig")
print(f"Created {output_csv} with {len(df_txt)} text files")
df_txt.head()
```
